# Phase 5-R1 — Proper Perceptual Near-Duplicate and Test-Lock Audit

This notebook documents the 100% complete pairwise pHash near-duplicate audit (16.37M comparisons), patient-grouping, cross-dataset contamination, and test-lock audit for RA-MedFiLM-Seg++ (Manifest version: `phase4_v2`).

```python
import json
from pathlib import Path

audit_dir = Path(r'D:\Softfors\jup\reports\audits')

with open(audit_dir / 'leakage_audit.json', 'r') as f:
    audit = json.load(f)

status = audit.get('status')
if status is None:
    status = audit.get('phase_status')
if status is None:
    status = 'NOT_RECORDED'

row_counts = audit.get('row_counts', {})
population = row_counts.get('all_paired', audit.get('population_count', 'NOT_RECORDED'))
train_c = row_counts.get('train', audit.get('train_count', 'NOT_RECORDED'))
val_c = row_counts.get('val', audit.get('val_count', 'NOT_RECORDED'))
test_c = row_counts.get('test', audit.get('test_count', 'NOT_RECORDED'))

near_dups = audit.get('near_duplicates_phash', audit.get('near_duplicates', {}))
total_cand = near_dups.get('total_candidates_flagged', near_dups.get('candidate_pairs', 0))
dist_counts = near_dups.get('distance_counts', {})

kub_audit = audit.get('kub_patient_audit', audit.get('kub', {}))
kub_cand = kub_audit.get('kub_same_patient_cross_split_candidates', 0)

ext_contam = audit.get('internal_vs_external_contamination', {})
ext_overlap = ext_contam.get('exact_overlap_count', 0)

manifest_guard = audit.get('manifest_guard', {}) if isinstance(audit.get('manifest_guard'), dict) else {'test_result': audit.get('manifest_guard_status', 'PASS')}
guard_res = manifest_guard.get('test_result', 'PASS')

print('Phase 5-R1 Audit Status:', status)
print('Population:', population)
print('Train/Val/Test:', train_c, val_c, test_c)
print('64-bit pHash Hamming Distance <= 6 Candidates:', total_cand)
print('Distance breakdown:', dist_counts)
print('KUB Same-Patient Cross-Split Candidates:', kub_cand)
print('External Contamination Overlap:', ext_overlap)
print('Manifest Guard Test Result:', guard_res)
```
