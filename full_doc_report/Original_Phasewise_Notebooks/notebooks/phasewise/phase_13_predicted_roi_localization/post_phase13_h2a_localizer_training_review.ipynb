## 24. Limitations

- **Adaptive development.** H2A was initiated after Phase 13 failed the gate; the selection rule was frozen before training and VAL is development data.
- **Epochs 1-30 carry no precision/coverage signal.** Proposals only appear from epoch 31, so the tuple's first five terms are zero there and selection falls back to `-val_loss`. The rule therefore cannot discriminate epochs 1-30 on precision; divergence from Phase-13 is confined to the post-epoch-31 regime.
- **Single seed** (42); no paired significance testing was performed and none is claimed.
- **Selection is not a causal proof.** A better-selected checkpoint does not establish that selection redesign is sufficient downstream.
- **No downstream segmentation was run** in this part, so no regional-integration conclusion follows.
- HD95-style fields are not part of this localizer-level contract.
- The comparison against Phase-13 uses the **frozen** Part-1C tables (no Phase-13 re-inference); it is descriptive and makes **no superiority or significance claim**.
