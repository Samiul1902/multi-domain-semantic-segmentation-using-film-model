## Visual Standards & Color Code Reference

All error maps in this project adhere to the unified standard:
- **Green / Cyan (TP):** True Positive detection/segmentation overlapping ground truth stones.
- **Red / Orange (FP):** False Positive prediction in kidney/ureter/bladder background.
- **Blue / Yellow (FN):** False Negative missed stone tissue.
- **Black / Transparent:** True Negative background tissue.
