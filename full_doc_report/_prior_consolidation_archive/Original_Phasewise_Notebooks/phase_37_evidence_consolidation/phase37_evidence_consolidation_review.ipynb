## 0. Locate the repository root and load Phase37 artifacts

```python
from pathlib import Path
import json

def find_root(start):
    p = Path(start).resolve()
    for c in [p, *p.parents]:
        if (c / 'reports' / 'audits' / 'phase37_evidence_consolidation_summary.json').exists():
            return c
    return Path.cwd()

ROOT = find_root(Path.cwd())
A = ROOT / 'reports' / 'audits'

def load(name):
    return json.loads((A / name).read_text(encoding='utf-8'))

SUMMARY = load('phase37_evidence_consolidation_summary.json')
INV = load('phase37_master_evidence_inventory.json')
CM = load('phase37_claim_evidence_matrix.json')
TR = load('phase37_table_registry.json')
FR = load('phase37_figure_registry.json')
LED = load('phase37_manuscript_results_ledger.json')
XR = load('phase37_cross_reference_map.json')
CA = load('phase37_contradiction_qualification_audit.json')
ACCESS = load('phase37_protected_population_access_audit.json')
GOV = load('phase37_governance_no_new_science.json')
GATE = load('phase37_acceptance_gate.json')
REG = load('phase37_frozen_artifact_registry.json')
print('Phase37 artifacts loaded from', A)
print('frozen Phase27-36 artifacts in registry:', REG['n_artifacts'])
```
